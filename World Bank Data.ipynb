```python
import urllib.request
import json
import pandas as pd
import matplotlib.pyplot as plt

def ambil_data_worldbank(kode_indikator, kode_negara, tahun_mulai, tahun_selesai):
    url = f"https://api.worldbank.org/v2/country/{kode_negara}/indicator/{kode_indikator}?date={tahun_mulai}:{tahun_selesai}&format=json&per_page=100"

    with urllib.request.urlopen(url) as response:
        data = json.loads(response.read())

    records = []
    for item in data[1]:
        if item["value"] is not None:
            records.append({
                "Tahun": int(item["date"]),
                "Nilai": item["value"],
                "Negara": item["country"]["value"]
            })

    return pd.DataFrame(records).sort_values("Tahun")

print("⏳ Mengambil data dari World Bank...")

gdp_singapore = ambil_data_worldbank(
    kode_indikator="NY.GDP.MKTP.CD",
    kode_negara="SG",
    tahun_mulai=1980,
    tahun_selesai=2025
)



print("=== GDP Singapore 1980-2023 ===")
print(gdp_singapore.tail(10))

plt.figure(figsize=(12, 5))
plt.plot(gdp_singapore["Tahun"], gdp_singapore["Nilai"] / 1e12,
         marker="o", color="steelblue", linewidth=2)
plt.title("GDP Singapore 1980-2025 (Triliun USD)")
plt.xlabel("Tahun")
plt.ylabel("GDP (Triliun USD)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import urllib.request
import json
import pandas as pd
import matplotlib.pyplot as plt

def ambil_data_worldbank(kode_indikator, kode_negara, tahun_mulai, tahun_selesai):
    url = f"https://api.worldbank.org/v2/country/{kode_negara}/indicator/{kode_indikator}?date={tahun_mulai}:{tahun_selesai}&format=json&per_page=100"

    with urllib.request.urlopen(url) as response:
        data = json.loads(response.read())

    records = []
    # Check if the response contains actual data records at index 1
    if len(data) > 1 and isinstance(data[1], list):
        for item in data[1]:
            if item["value"] is not None:
                records.append({
                    "Tahun": int(item["date"]),
                    "Nilai": item["value"],
                    "Negara": item["country"]["value"]
                })
    else:
        print(f"Warning: No data found for indicator '{kode_indikator}' in '{kode_negara}' for years {tahun_mulai}-{tahun_selesai}.")
        return pd.DataFrame(columns=["Tahun", "Nilai", "Negara"]) # Return empty DataFrame with expected columns

    return pd.DataFrame(records).sort_values("Tahun")

print("⏳ Mengambil data dari World Bank...")

gdp_japan = ambil_data_worldbank(
    kode_indikator="NY.GDP.MKTP.CD",
    kode_negara="JP",
    tahun_mulai=1980,
    tahun_selesai=2025
)

gdp_united_kingdom = ambil_data_worldbank(
    kode_indikator="NY.GDP.MKTP.CD",
    kode_negara="GB",
    tahun_mulai=1980,
    tahun_selesai=2025
)

gdp_germany = ambil_data_worldbank(
    kode_indikator="NY.GDP.MKTP.CD",
    kode_negara="DE",
    tahun_mulai=1980,
    tahun_selesai=2025
)

print("=== Education Japan, United Kingdom & Germany 1980-2023 ===")
print(gdp_japan.tail(10))
print(gdp_united_kingdom.tail(10))
print(gdp_germany.tail(10))

plt.figure(figsize=(12, 5))
plt.plot(gdp_japan["Tahun"], gdp_japan["Nilai"] / 1e+12,
         marker="o", color="steelblue", linewidth=2)
plt.plot(gdp_united_kingdom["Tahun"], gdp_united_kingdom["Nilai"] / 1e+12,
         marker="o", color="red", linewidth=2)
plt.plot(gdp_germany["Tahun"], gdp_germany["Nilai"] / 1e+12,
         marker="o", color="brown", linewidth=2)
plt.title("GDP Japan (Biru), United Kingdom (Merah), & Germany (Coklat) 1980-2025 (Triliun USD)")
plt.xlabel("Tahun")
plt.ylabel("GDP (Triliun USD)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import urllib.request
import json
import pandas as pd
import matplotlib.pyplot as plt

def ambil_data(indikator, negara, tahun_mulai=1970, tahun_selesai=2025):
    url = f"https://api.worldbank.org/v2/country/{negara}/indicator/{indikator}?date={tahun_mulai}:{tahun_selesai}&format=json&per_page=100"
    with urllib.request.urlopen(url) as r:
        data = json.loads(r.read())
    records = []

    if len(data) > 1 and isinstance(data[1], list):
        for item in data[1]:
            if item["value"] is not None:
                records.append({"Tahun": int(item["date"]), "Nilai": item["value"]})
    else:
        print(f"Warning: No data found or unexpected format for {negara} with indicator {indikator}.")
        return pd.DataFrame(columns=["Tahun", "Nilai"])
    return pd.DataFrame(records).sort_values("Tahun")

MODELS = [
    "google/gemma-3-4b-it:free",
    "mistralai/devstral-small:free",
    "meta-llama/llama-3.2-3b-instruct:free",
]

def tanya_ai(prompt):
    url = "https://openrouter.ai/api/v1/chat/completions"
    headers = {
        "Content-Type": "application/json",
        "Authorization": "Bearer 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"
    }

    for model in MODELS:
        try:
            payload = {
                "model": model,
                "messages": [{"role": "user", "content": prompt}]
            }
            data = json.dumps(payload).encode("utf-8")
            req = urllib.request.Request(url, data=data, headers=headers)
            with urllib.request.urlopen(req) as r:
                result = json.loads(r.read())
            print(f"✅ Pakai model: {model}")
            return result["choices"][0]["message"]["content"]
        except:
            print(f"❌ {model} gagal, coba berikutnya...")
            continue

    return "Semua model gagal. Coba lagi nanti."
    headers = {
        "Content-Type": "application/json",
        "Authorization": "Bearer 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"
    }
    data = json.dumps(payload).encode("utf-8")
    req = urllib.request.Request(url, data=data, headers=headers)
    with urllib.request.urlopen(req) as r:
        result = json.loads(r.read())
    return result["choices"][0]["message"]["content"]

print("=" * 60)
print("     ADB COUNTRY INTELLIGENCE TOOL")
print("=" * 60)

jumlah = int(input("Mau analisis berapa negara? "))
negara_list = []

for i in range(jumlah):
    nama = input(f"\nNama negara {i+1}: ")
    kode = input(f"Kode negara {i+1} (2 huruf): ").upper()
    negara_list.append((nama, kode))

for nama_negara, kode_negara in negara_list:
    print(f"\n⏳ Mengambil data {nama_negara} dari World Bank...\n")

    gdp          = ambil_data("NY.GDP.MKTP.CD", kode_negara)
    pendidikan = ambil_data("SE.XPD.TOTL.GD.ZS", kode_negara)

    fig, axes = plt.subplots(1, 2, figsize=(20, 4))
    fig.suptitle(f"Economic Profile: {nama_negara}", fontsize=14, fontweight="bold")

    axes[0].plot(gdp["Tahun"], gdp["Nilai"]/1e9, color="steelblue", marker="o", markersize=3)
    axes[0].set_title("GDP (Miliar USD)")
    axes[0].set_xlabel("Tahun")
    axes[0].grid(True, alpha=0.3)

    axes[1].plot(pendidikan["Tahun"], pendidikan["Nilai"], color="purple", marker="o", markersize=3)
    axes[1].set_title("Education Spending (% GDP)")
    axes[1].set_xlabel("Tahun")
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    gdp_terbaru = gdp.iloc[-1]["Nilai"] / 1e9
    gdp_terlama = gdp.iloc[0]["Nilai"] / 1e9
    gdp_growth  = ((gdp_terbaru - gdp_terlama) / gdp_terlama) * 100

    print(f"\n📊 RINGKASAN DATA {nama_negara.upper()}")
    print(f"GDP terbaru:  ${gdp_terbaru:.1f} Miliar")
    print(f"GDP growth:   {gdp_growth:.1f}% sejak {gdp.iloc[0]['Tahun']}")
    if not pendidikan.empty:
        print(f"Education rate: {pendidikan.iloc[-1]['Nilai']:.1f}%")

    print(f"\n🤖 AI lagi nulis analisis {nama_negara}...\n")

    prompt = f"""
    Kamu adalah senior economist di ADB.
    Tulis analisis singkat (1 paragraf) tentang {nama_negara}:
    - GDP terbaru: ${gdp_terbaru:.1f} Miliar USD
    - GDP growth sejak {gdp.iloc[0]['Tahun']}: {gdp_growth:.1f}%
    - Education rate: {pendidikan.iloc[-1]['Nilai'] if not pendidikan.empty else 'N/A'}%
    Fokus: trajectory ekonomi, tantangan utama, rekomendasi investasi ADB.
    Bahasa Indonesia, formal tapi accessible.
    """

    analisis = tanya_ai(prompt)
    print("=" * 60)
    print(f"📋 COUNTRY BRIEF: {nama_negara.upper()}")
    print("=" * 60)
    print(analisis)
    print("=" * 60)
```
